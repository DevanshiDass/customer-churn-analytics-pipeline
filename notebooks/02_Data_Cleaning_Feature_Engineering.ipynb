```python
# ==========================================
# STEP 3: DATA CLEANING & FEATURE ENGINEERING
# ==========================================

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv(r"C:\Users\usdas\Desktop\DEVANSHI ALLLLL\PROJECTS\customer-churn-analysis\data\raw\churn_data.csv")

# -------------------------------
# 1. FIX DATA TYPES
# -------------------------------

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# -------------------------------
# 2. HANDLE MISSING VALUES
# -------------------------------

print(df.isnull().sum())

df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())

# -------------------------------
# 3. DROP UNNECESSARY COLUMNS
# -------------------------------

df = df.drop('customerID', axis=1)

# -------------------------------
# 4. CONVERT TARGET VARIABLE
# -------------------------------

df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# -------------------------------
# 5. ENCODE GENDER (FIXED)
# -------------------------------

df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})

# -------------------------------
# 6. ENCODE OTHER BINARY FEATURES
# -------------------------------

binary_cols = [
    'Partner', 'Dependents',
    'PhoneService', 'PaperlessBilling'
]

for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

# -------------------------------
# 7. ENCODE MULTI-CATEGORY FEATURES
# -------------------------------

multi_cols = [
    'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup',
    'DeviceProtection', 'TechSupport',
    'StreamingTV', 'StreamingMovies',
    'Contract', 'PaymentMethod'
]

df = pd.get_dummies(df, columns=multi_cols, drop_first=True)

# -------------------------------
# 8. FEATURE ENGINEERING
# -------------------------------

df['AvgMonthlySpend'] = df['TotalCharges'] / (df['tenure'] + 1)
df['HighValueCustomer'] = np.where(df['MonthlyCharges'] > 70, 1, 0)

# -------------------------------
# 9. FEATURE SCALING
# -------------------------------

scaler = StandardScaler()
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges', 'AvgMonthlySpend']
df[num_cols] = scaler.fit_transform(df[num_cols])

# -------------------------------
# 10. FINAL DATA CHECK
# -------------------------------

print(df.head())
print(df.shape)
print("Total missing values:", df.isnull().sum().sum())

# -------------------------------
# 11. SAVE CLEAN DATA
# -------------------------------

df.to_csv(
    r"C:\Users\usdas\Desktop\DEVANSHI ALLLLL\PROJECTS\customer-churn-analysis\data\processed\churn_cleaned.csv",
    index=False
)

print("✅ Data cleaning & feature engineering completed.")
```
